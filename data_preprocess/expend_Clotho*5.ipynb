```python
# gj
import pandas as pd
import glob

# 读取 parquet 文件
# 匹配目录下所有 parquet 文件
files = glob.glob("/mnt/datasets/liyaqi/Clotho/*.csv", recursive=True)

# 逐个读取并拼接
df = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)

display(df)
```

```python
import os

wavs = df["file_name"].tolist()
for wav in wavs:

    # 原始文件路径
    src_path = "/mnt/datasets/liyaqi/Clotho/development+evaluation/"+wav
    # 输出随机文件路径
    dst_path = "/mnt/datasets/ExpandedClotho/wavs/"

    # 获取原始文件大小（字节数）
    size = os.path.getsize(src_path)

    for i in range(5):
        # 生成同样大小的随机字节
        random_data = os.urandom(size)
        name, ext = os.path.splitext(wav)
        new_name = f"{name}({i}){ext}"
        # 写入新文件
        with open(dst_path+new_name, "wb") as f:
            f.write(random_data)

```

```python
newdf=pd.DataFrame(columns=df.columns)
import random
import string

for index, row in df.iterrows():
    filename=row["file_name"]
    captions=[row["caption_1"],row["caption_2"],row["caption_3"],row["caption_4"],row["caption_5"],]

    for i in range(5):
        name, ext = os.path.splitext(filename)
        new_name = f"{name}({i}){ext}"
        new_captions=[]
        for j in range(5):
            new_captions+=[''.join(random.choices(string.ascii_letters, k=len(captions[j])))]
        newdf.loc[len(newdf)]=[new_name]+new_captions

display(newdf)
```

```python
newdf.to_csv('/mnt/datasets/ExpandedClotho/ExpandedClotho.csv', index=False)
```

```python
images = newdf["file_name"].tolist()
captions=[newdf["caption_1"].tolist(),newdf["caption_2"].tolist(),newdf["caption_3"].tolist(),newdf["caption_4"].tolist(),newdf["caption_5"].tolist(),]
res = {"image_path":[],"image_shape":[],"image_tensor_size":[],"image_file_size":[],"compression_rate":[],"image_caption":[],"image_caption_size":[]}

image_dir = "/mnt/datasets/ExpandedClotho/wavs"

for i in range(len(images)):
    if i % round(len(images)/4) == 0:
        print(i/len(images))

    image_path = images[i]
    image_full_path = os.path.join(image_dir, image_path)
    for j in range(5):
        res["image_path"].append(image_path)
        res["image_shape"].append(0)
        # res["image_tensor_size"].append(image_tensor_size)
        res["image_tensor_size"].append(0)
        image_file_size = os.path.getsize(image_full_path)
        res["image_file_size"].append(image_file_size)
        # res["compression_rate"].append(f"{image_tensor_size/image_file_size:0.2f}")
        res["compression_rate"].append(0)
        image_caption = captions[j][i]
        res["image_caption"].append(image_caption)
        image_caption_size = len(image_caption)
        res["image_caption_size"].append(image_caption_size)

out = pd.DataFrame(res)
# out.to_csv("MSRVTT_sample.csv",index=False)

display(out)
```

```python
# 提取数字并转换为整数类型
out['numeric_id'] = out['image_path']

# 按数字升序排序
sorted_df = out.sort_values(by='numeric_id', ascending=True)

# 生成新CSV文件（删除辅助列）
sorted_df.drop(columns=['numeric_id']).to_csv('/home/llm/experiment/dataset_Analysis_Modeling/processed_datasets/ExpandedClotho_sort.csv', index=False)
```
